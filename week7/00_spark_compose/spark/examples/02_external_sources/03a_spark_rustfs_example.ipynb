```python
from pyspark.sql import SparkSession
```

```python
accessKeyId='dataops'
secretAccessKey='Ankara06'

# create a SparkSession
spark = SparkSession.builder \
.appName("Spark Example RustFS") \
.master("local[2]") \
.config("spark.jars.packages", "org.apache.hadoop:hadoop-aws:3.4.2") \
.config("spark.hadoop.fs.s3a.access.key", accessKeyId) \
.config("spark.hadoop.fs.s3a.secret.key", secretAccessKey) \
.config("spark.hadoop.fs.s3a.path.style.access", True) \
.config("spark.hadoop.fs.s3a.endpoint", "http://rustfs:9000") \
.getOrCreate()
```

```python
df = spark.read.format('csv').option('header','true') \
.load('s3a://datasets/iris.csv')
```

```python
df.show(5)
```

```python
df.write.mode('overwrite').save('s3a://datasets/iris_parquet')
```

```python
spark.read \
.load('s3a://datasets/iris_parquet').show(3)
```

```python
spark.stop()
```
