```python
from pyspark.sql import SparkSession, functions as F
```

```python
spark = SparkSession.builder \
.appName("Define a Custom Schema") \
.master("local[2]") \
.getOrCreate()
```

```python
df = spark.read.format("csv") \
.option("sep", ",") \
.option("header", True) \
.option("inferSchema", True) \
.load("file:///opt/examples/datasets/OnlineRetail2.csv")
```

```python
print(df.count())
```

```python
df.limit(3).toPandas()
```

```python
df.printSchema()
```

```python
from pyspark.sql.types import StructField, StructType, StringType, IntegerType, DoubleType, FloatType, TimestampType
schema1 = StructType(
[
StructField('InvoiceNo', StringType(), True), 
StructField('StockCode', StringType(), True), 
StructField('Description', StringType(), True), 
StructField('Quantity', IntegerType(), True), 
StructField('InvoiceDate', TimestampType(), True), 
StructField('UnitPrice', FloatType(), True), 
StructField('CustomerID', DoubleType(), True), 
StructField('Country', StringType(), True)
]

)
```

```python
df2 = spark.read.format("csv") \
.option("sep", ",") \
.option("header", True) \
.schema(schema1) \
.load("file:///opt/examples/datasets/OnlineRetail2.csv")
```

```python
df2.limit(3).toPandas()
```

```python
df2.printSchema()
```

```python
sql_way_schema = """InvoiceNo string,
StockCode string,
Description string, 
Quantity integer, 
InvoiceDate timestamp,
UnitPrice float,
CustomerID double,
Country string"""
```

```python
df3 = spark.read.format("csv") \
.option("sep", ",") \
.option("header", True) \
.schema(sql_way_schema) \
.load("file:///opt/examples/datasets/OnlineRetail2.csv")
```

```python
df3.show(3)
```

```python
df3.printSchema()
```
