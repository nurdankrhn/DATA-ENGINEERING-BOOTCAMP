```python
from pyspark.sql import SparkSession, functions as F
from pyspark.sql.types import *
```

```python
spark = SparkSession.builder \
.appName("Modify Schema") \
.master("local[2]") \
.getOrCreate()
```

```python
! wget -O /opt/examples/datasets/Hotel_Reviews.csv.gz \
https://github.com/erkansirin78/datasets/raw/master/Hotel_Reviews.csv.gz
```

```python
! ls -l /opt/examples/datasets/
```

```python
df = spark.read.option("header", True) \
.option("inferSchema", True) \
.csv("file:///opt/examples/datasets/Hotel_Reviews.csv.gz")
```

```python
df.limit(10).toPandas()
```

```python
df.count()
```

```python
df.printSchema()
```

```python
df2 = df.withColumn("Review_Date", F.to_date("Review_Date", "M/d/yyyy"))
```

```python
df2.limit(1).toPandas()
```

```python
df2.printSchema()
```

```python
df3 = df2.withColumn("Tags", 
                     F.split(F.col("Tags"), ",")
                     .cast(ArrayType(StringType()))) \
.withColumn("lat", F.col("lat").cast(FloatType())) \
.withColumn("lng", F.col("lng").cast(FloatType()))
```

```python
df3.limit(1).toPandas()
```

```python
df3.printSchema()
```
